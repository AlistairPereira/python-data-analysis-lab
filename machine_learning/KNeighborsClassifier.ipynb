```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

```python
df = pd.read_csv("D:\git\python-data-analysis-lab\datasets\Social_Network_Ads.csv")
df.head()
```

```python
sns.scatterplot(data=df, x="Age", y ="EstimatedSalary",hue="Gender")
plt.show()
```

```python
x=df.iloc[:,[2,3]]
x
```

```python
y=df.iloc[:,-1]
y
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=0.25, random_state=0)
```

```python
len(xtrain),len(xtest),len(ytrain),len(ytest)
```

```python
knc = KNeighborsClassifier()
```

```python
knc.fit(xtrain,ytrain)
```

```python
ypred = knc.predict(xtest)
ypred
```

```python
xpred = knc.predict(xtrain)
xpred
```

```python
train_ac = accuracy_score(ytrain,xpred)
test_ac = accuracy_score(ytest,ypred)
```

```python
train_ac, test_ac
```

```python
cm = confusion_matrix(ytest,ypred)
cm
```

```python
cr = classification_report(ytest,ypred)
print(f" classification report : {cr}")
```

```python
trainac=[]
testac=[]
for i in range(1,31):
    knc = KNeighborsClassifier(n_neighbors=i)
    knc.fit(xtrain,ytrain)
    train = knc.score(xtrain,ytrain)
    test = knc.score(xtest,ytest)
    trainac.append(train)
    testac.append(test)
```

```python
plt.plot(range(1,31),trainac,color="blue")
plt.plot(range(1,31),testac,color="red")
plt.grid()
plt.show()
```
