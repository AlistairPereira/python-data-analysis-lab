```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv("D:\git\python-data-analysis-lab\datasets\Scores.csv")
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
def grades(score):
    if score >= 75:
        return "A"
    elif score >= 60 and score < 75:
        return "B"
    elif score >= 35 and score < 60:
        return "C"
    else:
        return "D"
```

```python
df['Grades'] = df['Scores'].apply(grades)
```

```python
df.head()
```

```python
sns.scatterplot(data =df, x="Hours",y="Scores")
plt.show()
```

```python
x=df[['Hours']]
y=df[['Scores']]
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x,y,test_size=0.2, random_state=1)
```

```python
len(xtrain), len(ytrain), len(xtest), len(ytest)
```

```python
linreg =LinearRegression()
```

```python
linreg.fit(xtrain, ytrain)
```

```python
ypred = linreg.predict(xtest)
ypred
```

```python
ytest
```

```python
xpred = linreg.predict(xtrain)
xpred
```

```python
ytrain
```

```python
plt.scatter(xtest,ytest, color ='blue')
plt.plot(xtest, ypred, color="red")
plt.grid()
plt.show()
```

```python
plt.scatter(xtrain,ytrain, color ='blue')
plt.plot(xtrain, xpred, color="red")
plt.grid()
plt.show()
```

```python
mae = mean_absolute_error(ytest,ypred)
mae
```

```python
mse = mean_squared_error(ytest,ypred)
mse
```

```python
r2=r2_score(ytest,ypred)
r2
```
